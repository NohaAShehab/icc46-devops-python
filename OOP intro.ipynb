```python
# class --> properties (), functions / methods

# customize object while creation
class Employee():
    def __init__(self, id, name, salary): # function ,,,
        # id, name , salary --> instance variables,
        # values of these variables depends on caller instance
        self.id =id
        self.name =name
        self.salary = salary

    # instance method -> function depends caller instance
    def printEmp(self):# self access object
        print(f"{self.id}, {self.name}, {self.salary}")

e1 = Employee(3,'rrr',33)
e2 = Employee(4, 'new', 344)
e1.printEmp()
e2.printEmp()



```

```python
####################################
class Student:
    def __init__(self, name):
        self.name = name

    def printName(self):
        print(f'student namr ={self.name}')

    @staticmethod
    def formatName(name):
        print(f'formatted name {name.upper()}')

s = Student("noha")
s.printName()
```
