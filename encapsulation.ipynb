```python
## why we need to create private , protected members

# setters and getters
class Student:
    def __init__(self, name, id , grade):
        self.name= name
        self.id = id
        self.__grade = grade  # acts as private member

    def set_grade(self,grade):
        if isinstance(grade, int) or isinstance(grade, float) and grade>0:
            self.__grade =grade
        else:
            raise ValueError("Grade must be number greater than zero")

    def get_grade(self):
        return self.__grade


s = Student("Ahmed", 24, 90)
print(s.get_grade())
s.set_grade(100)
print(s.get_grade())

```
