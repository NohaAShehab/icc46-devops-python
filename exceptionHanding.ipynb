```python
# I need to run some lines only if no errors happeded ?
try:
    num1 = int(input("please enter num1 "))
    num2 = int(input("please enter num2 "))
    res = num1 + num2
    print(res)
except ValueError as ve :
    print(f"--- error happened while casting numbers {ve}")
except NameError as ne:
    print(f"--- variable not found {ne}")
except Exception as e:
    print(e)
else:
    "this block will be called only if there is no problem"
    print("---- code executed successfully, no errors ")
```

```python
# finally block?

# I need to run some lines only if no errors happeded ?
try:
    num1 = int(input("please enter num1 "))
    num2 = int(input("please enter num2 "))
    res = num1 + num2
    print(res)
except ValueError as ve :
    print(f"--- error happened while casting numbers {ve}")
except NameError as ne:
    print(f"--- variable not found {ne}")
except Exception as e:
    print(e)
else:
    "this block will be called only if there is no problem"
    print("---- code executed successfully, no errors ")
finally:
    print("---- this block run always")
print("----------------------------------- me also ")
```

```python
def askforInt():
    try:
        num = int(input("please enter number : "))
    except Exception as e:
        print(e)
        return False
    else:
        return num
    finally:
        """ execution of finally block , precedes --> return ? """
        print('----- function ended -----')

print(askforInt())
```
