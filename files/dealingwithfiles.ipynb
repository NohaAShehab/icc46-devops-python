<h3 style='color:green'>Read Data from file </h3>

```python
fileObj = open("users.txt")
print(fileObj)

```

```python
try:
    fileObj = open("users.txt") # assume open file for reading
    print(fileObj)
except Exception as e:
    print(e)
```

```python
try:
    fileObj2 = open("users.txt", "r") # assume open file for reading
    print(fileObj2)
except Exception as e:
    print(e)
else:
    data = fileObj2.read() # read file content from beginning of the file ?
    print(data)
    fileObj2.close()
```

```python
# read file into lines
try:
    fileObj = open("users.txt") # assume open file for reading
    print(fileObj)
except Exception as e:
    print(e)
else:
    lines = fileObj.readlines()
    print(lines)
    fileObj.close()
```

```python
# move to byte 10 then read
try:
    fileObj = open("users.txt") # assume open file for reading
    print(fileObj)
except Exception as e:
    print(e)
else:
    fileObj.seek(10)
    print(fileObj.read())
    fileObj.close()
```

<h3 style='color:pink'>Save Data to a file  </h3>
when you try to open file with mode w --> if file doesn't exist, will create it
if file exists ? open file for writing starting from the beginning of the file ?
remove old content ?
===> this in case if we have permission to do writing on the file


```python
try:
    fileobj = open("mycv", "w")
    print(fileObj)
except Exception as e:
    print(e)
else:
    print("-- file opened ")
    fileObj.close()
```

```python
try:
    fileObj = open("mycv", "w")
    print(fileObj)
except Exception as e:
    print(e)
else:
    print("-- file opened ")
    fileObj.write("hello form mycv\n")
    fileObj.write("python is easy\n")
    fileObj.write("abc")
    fileObj.close()
```

```python
try:
    fileObj = open("mycv.txt", "w")
    print(fileObj)
except Exception as e:
    print(e)
else:
    print("-- file opened ")
    fileObj.seek(20)
    fileObj.writelines(["new\n", "word\n", "line\n"])
    fileObj.close()
```

<h3 style='color:orange'>Append Data to a file  </h3>
when you try to open file with mode a --> if file doesn't exist, will create it
if file exists ? open file for writing starting from the end of the file ?
keep old content ?
===> this in case if we have permission to do writing on the file


```python
try:
    fileObj = open("mycv.txt", "a")
    print(fileObj)
except Exception as e:
    print(e)
else:
    print("-- file opened ")
    fileObj.seek(10)
    fileObj.write("----------\n")
    fileObj.writelines(["new\n", "word\n", "line\n"])
    fileObj.close()
```
