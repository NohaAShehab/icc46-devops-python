# functions with unknown number of parameters

```python
def processCVs(cvs):
    print(cvs)


processCVs("wee")

processCVs("afsdf", "afaff", "asdfasf")
```

```python
# function --> accept zero or argument

def processCVs(*cvs):  # * --> zero or more arguments
    print(cvs)


processCVs("wee")

processCVs("afsdf", "afaff", "asdfasf")
```

```python
## You must do the type validation on your own

def sumnum(num1: int, num2: int):
    if type(num1)==int and type(num2)==int:
        print(num1 + num2)
        return  num1 + num2

    print("--- num1, num2 must be integers ")
    return False


sumnum(3,4)
```

```python
def sumnum(num1: int, num2: int):
    if isinstance(num1, int) and isinstance(num2, int):
        print(num1 + num2)
        return  num1 + num2

    print("--- num1, num2 must be integers ")
    return False


sumnum(3,4)
```
