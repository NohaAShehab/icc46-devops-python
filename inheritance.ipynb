```python
## modify behaviour of child class
class Parent:
    def __init__(self, name):
        self.name = name

    def printName(self):
        print(self.name)

    def info(self):
        print("--- parent info")

class Child(Parent):
    def __init__(self, name,age):
        self.age = age
        super().__init__(name)

    def printAge(self):
        print(f"age ={self.age}")

c= Child('Ahmed', 3)
print(c.__dict__)
c.printName()
c.printAge()
c.info()

```

```python
# i need to modify info function
## modify behaviour of child class
class Parent:
    def __init__(self, name):
        self.name = name

    def printName(self):
        print(self.name)

    def info(self):
        print("--- parent info")

class Child(Parent):
    def __init__(self, name,age):
        self.age = age
        super().__init__(name)

    def printAge(self):
        print(f"age ={self.age}")

     # override --> 2 classes have inheritance relation have 2 functions with the same name , but in parent do something, child do other thing
    def info(self):
        super().info()
        print("--- this is the info of the child ")

c= Child('Ahmed', 3)
print(c.__dict__)
c.printName()
c.printAge()
c.info()

```

```python
# overloading

class Student:
    def __init__(self, name='abc'):
        self.name = name
    def printinfo(self, name):
        print("bye", name)

    def printinfo(self):
        print("hi")


s = Student()
s.printinfo()
s.printinfo("Ahmed")

# overload is not applied explicitly in python.
```

```python
# multiple inheritance
class A:
    def __init__(self):
        print("-- object A created --")

class B:
    def __init__(self):
        print("-- object B created ")

class C(A,B):
    def __init__(self):
        super().__init__() # this will call first parent ...
        B.__init__(self)  # call it explicitly
        print("c object is created ")

c= C()
print(isinstance(c, A), isinstance(c, B))
```

```python
# multiple inheritance

class Base:
    def __init__(self):
        print("base is called ")
class A(Base):
    def __init__(self):
        print("-- object A created --")

class B(Base):
    def __init__(self):
        print("-- object B created ")

class C(A,B):
    def __init__(self):
        super().__init__() # this will call first parent ...

        print("c object is created ")

c= C()
print(isinstance(c, A), isinstance(c, B))
```
