# conversion between types

```python
year = 2025 # int
print(year, type(year))
year = str(year)  # type casting
print(year, type(year))
```

```python
grade = "100"
print(grade, type(grade))

grade = int(grade)
print(grade, type(grade))

```

```python
name = '10ali'
print(name, type(name))
name = int(name)
print(name, type(name))
```
